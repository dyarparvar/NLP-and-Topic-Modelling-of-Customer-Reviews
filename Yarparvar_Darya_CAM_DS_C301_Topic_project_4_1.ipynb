## Reproducibility

Function to establish reproducibility by fixing the seed

```python
def set_seed(seed=42):
    # Python randomness
    random.seed(seed)
    # Python hash randomness
    os.environ["PYTHONHASHSEED"] = str(seed)

    # NumPy randomness
    np.random.seed(seed)

    # UMAP stability
    umap_model = UMAP(
        n_neighbors=15,
        n_components=5,
        min_dist=0.0,
        metric="cosine",
        random_state=seed,
        n_jobs=1 # set n_jobs=1 to ensure the GPU/CPU doesn't reorder calculations
    )

    # Torch and T4 GPU
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    return umap_model
```

```python
def clean_location_names(data, location_name):
    # Create a copy
    data = data.copy()

    # Convert to lowercase for consistency
    data[location_name] = data[location_name].str.lower()

    # Remove leading/trailing whitespace and non-breaking spaces
    data[location_name] = data[location_name].str.strip()
    data[location_name] = data[location_name].str.replace("\xa0", " ", regex=False)

    # Replace multiple spaces with single space
    data[location_name] = data[location_name].str.replace(r"\s+", " ", regex=True)

    # Remove hyphens and text before location name (for numbered entries)
    # e.g., '209 - Slagelse, Jernbanegade' -> 'slagelse jernbanegade'
    data[location_name] = data[location_name].str.replace(r"^\d+ - ", "", regex=True)

    # Remove " - " separators
    data[location_name] = data[location_name].str.replace(" - ", " ", regex=False)

    # Remove "ID XXX" suffixes
    data[location_name] = data[location_name].str.replace(r", id \d+$", "", regex=True)

    return data
```

`string.punctuation` only removes the standard ASCII apostrophe ', not the curly ones!

```python
# Function to remove punctuation
def remove_irr_char(text):

    # Remove emojis and Non-ASCII:
    text = re.sub(r'[^\x00-\x7F]+', ' ', text)

    # Replace newlines/tabs with spaces and collapse multiples
    text = " ".join(text.split())

    # Define characters to remove: punctuation + numbers
    chars_to_remove = string.punctuation + string.digits
    translator = str.maketrans("", "", chars_to_remove)

    # Translate the text
    cleaned_text = text.translate(translator)

    return " ".join(cleaned_text.split())
```

```python
# Use a regular expression to remove most standard punctuation
processed_data_tp["review_processed"] = [remove_irr_char(review) for review in processed_data_tp["review_processed"]]
processed_data_g["review_processed"] = [remove_irr_char(review) for review in processed_data_g["review_processed"]]
```
